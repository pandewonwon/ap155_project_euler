```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
g = 9.8 #[m/s^2]
l = 0.1
theta = 179*np.pi/180

def func(x, yvals):
    y0,y1 = yvals
    f0 = y1
    f1 = -g*np.sin(y0)/l
    return np.array([f0,f1])
```

```python
def rk4_gen(func,a,b,n,yinits):
    h = (b-a)/(n-1)
    xs = a + np.arange(n)*h
    ys = np.zeros((n,yinits.size))

    yvals = np.copy(yinits)
    for j,x in enumerate(xs):
        ys[j,:] = yvals
        
        k0 = h*func(x,yvals)
        k1 = h*func(x+h/2, yvals+k0/2)
        k2 = h*func(x+h/2, yvals+k1/2)
        k3 = h*func(x+h, yvals+k2)
        yvals += (k0+2*k1+2*k2+k3)/6
        
    return xs,ys
```

```python
a,b,n = 0,10,1000
yinits = np.array([theta, 0])
xs,ys = rk4_gen(func,a,b,n,yinits)
```

```python
plt.plot(xs,ys[:,0])
```
