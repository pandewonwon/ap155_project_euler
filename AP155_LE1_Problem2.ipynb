```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#assigning values
y=1
y_dprime=0.1
y_tprime=100
epsilon=2.2*(10**-16)

h=np.logspace(-4,-1,200)
```

```python
#forward difference error
error_fd=((h*y_dprime)/2)+((2*y*epsilon)/h)

#central difference error
error_cd=(((h**2)*y_tprime)/6)+((y*epsilon)/h)
```

```python
#Graphing
fig,ax=plt.subplots()

ax.loglog(h,error_fd,label="Forward Difference error")
ax.loglog(h,error_cd,label="Central Difference error")
ax.legend()
ax.set(
    xlabel="h values",
    ylabel="error"
)
ax.set_title("Error comparison for forward and central differences")
```

Oddly, the generated graph looks different as to the error comparison graph (Fig.3.2) from Gezerlis. What our graph is currently showing is only the right part of the Gezerlis graph or simply, the truncation/approximation dominated side. Specifically, we're looking at:

ϵ= h|f(x)|/2 for forward difference and

ϵ=h^2|f'''(x)|/6 for central difference

What we're also seeing is an intersection within the truncation side. It is expected that central difference should be superior as its error is expected to be smaller. This is not reflective of what we see in our graph. In our graph, the central difference error is rapidly increasing compared to the forward difference error, that at some h, it completely overpasses the forward difference error. This happens when our function and its derivatives have very large f'''(x) and extremely small f''(x). Additionally, the h^2 in the central difference error implies that the error is rapidly decreasing for small h and rapidly increasing has h increases.
