```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#x-axis points
N=100 #number of evenly spaced points in the x-axis
x=np.linspace(0.5,1.5,N)
```

```python
y=(x**6)+0.1*np.log(abs(1+3*(1-x)))

fig,ax = plt.subplots()

ax.plot(x,y)
ax.set(
    xlabel="x values",
    ylabel="f(x)"
)
ax.set_title("f(x) for N=100")
ax.grid()
```

Based on the graph, we are seeing a small dip in between 1.2<x<1.4.
Studying the given function, the second term is undefined at x=4/3. At that value, the second term becomes 0.1*log(0) which is undefined but more specifically, analytically approaches -∞ as x approaches 4/3. To see this more clearly, let's use a much finer grid.

```python
N=30000000
x=np.linspace(0.5,1.5,N)

y=(x**6)+0.1*np.log(abs(1+3*(1-x)))

fig,ax=plt.subplots()
ax.plot(x,y)
ax.set(
    xlabel="x values",
    ylabel="f(x)"
)
ax.set_title("f(x) for N=30000000 (finer grid)")
ax.grid()
```

Using a smaller or much finer grid, the dip becomes more apparent. This goes to show that we cannot simply use any grid granularity we feel like using. Say, for smaller N, the dip is isn't noticeable. It could easily be disregarded as a "lack of smoothness" in the graph because it is being evaluated at very few numbers. Little did we know, that the first term of the function (x^6) dominates the graph. We see here that by significantly increasing N, we are evaluating the function at values closer to the singularity and the analytical behavior of the function slowly shows up
