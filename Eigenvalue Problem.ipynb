```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def ham(x_pos, pot_array):
    N=len(x_pos)
    dx=x_pos[1]-x_pos[0]
    pot=np.diag(pot_array)
    kin=-(np.eye(N,N,k=+1)+np.eye(N,N,k=-1)-2*np.eye(N,N,k=0))/(dx**2)/2
    
    ham_matrix=pot+kin

    return ham_matrix
```

```python
N=1000
omega=0.9
x_pos=np.linspace(-10,10,N) #space is from -10<x<10
pot_array=0.5*(omega**2)*(x_pos**2) #harmonic oscillator potential

harmonic_oscillator=ham(x_pos,pot_array)
ho_eigvals,ho_eigvecs=np.linalg.eigh(harmonic_oscillator)
```

```python
for i in range(10):
    plt.plot(x_pos,ho_eigvals[i]+ho_eigvecs[:,i])

plt.plot(x_pos,pot_array,"k--")
plt.ylim(0,10)
```

```python
plt.plot(x_pos,ho_eigvals)
```

```python
def ham_finitesqwell(x_pos, pot_array_finitesqwell):
    N=len(x_pos)
    dx=x_pos[1]-x_pos[0]
    pot=np.diag(pot_array_finitesqwell)
    kin=-(np.eye(N,N,k=+1)+np.eye(N,N,k=-1)-2*np.eye(N,N,k=0))/(dx**2)/2
    
    ham_matrix=pot+kin

    return ham_matrix
```

```python
N=100
a= 3 #width of the well
v_0= 9 #depth of the well
x_pos=np.linspace(-10,10,N) #space is from -10<x<10
pot_array_finitesqwell=(v_0*(-a>x_pos)+v_0*(x_pos>a)) #finite square well potential

finite_sqwell=ham_finitesqwell(x_pos,pot_array_finitesqwell)
fsw_eigvals,fsw_eigvecs=np.linalg.eigh(finite_sqwell)
```

```python
plt.plot(x_pos,pot_array_finitesqwell)
```

```python
for i in range(100):
    plt.plot(x_pos,fsw_eigvals[i]+fsw_eigvecs[:,i])

plt.plot(x_pos,pot_array_finitesqwell,"k--")
plt.ylim(0,10)
```

```python
zero_eigvec=fsw_eigvecs[:,80]

plt.plot(x_pos,zero_eigvec)
```

```python
plt.plot(fsw_eigvals[:100],"o")
```

```python
#DOUBLE FINITE SQUARE WELL
```

```python
def ham_doublesqwell(x_pos, pot_array_doublesqwell):
    N=len(x_pos)
    dx=x_pos[1]-x_pos[0]
    pot=np.diag(pot_array_doublesqwell)
    kin=-(np.eye(N,N,k=+1)+np.eye(N,N,k=-1)-2*np.eye(N,N,k=0))/(dx**2)/2
    
    ham_matrix=pot+kin

    return ham_matrix
```

```python
N=100
a= 3 #width of the well
v_0= 15 #depth of the well
d= 10 #distance between wells
x_pos=np.linspace(-10,10,N) #space is from -10<x<10
pot_array_doublesqwell= v_0*((-d/2)-a>x_pos)+v_0*(x_pos>(-d/2))+v_0*((d/2)+a<x_pos)+v_0*(x_pos<(d/2)) #double finite square well potential

double_sqwell=ham_doublesqwell(x_pos,pot_array_doublesqwell)
dsw_eigvals,dsw_eigvecs=np.linalg.eigh(double_sqwell)
```

```python
fig,ax=plt.subplots()

plt.plot(x_pos,pot_array_doublesqwell)
ax.set_xlabel('x')
ax.set_ylabel('V(x)')
ax.set_title('Double Finite Square Well Potential')
plt.savefig("EVProblemPotential")
```

```python
fig,ax=plt.subplots()

for i in range(10):
    plt.plot(x_pos,dsw_eigvals[i]+dsw_eigvecs[:,i])

plt.plot(x_pos,pot_array_doublesqwell,"k--")
plt.ylim(13,31)
ax.set_xlabel('x')
ax.set_ylabel('Energy')
ax.set_title('Solutions to the Schrodinger Eq. for the double finite square well')
plt.savefig('EVProblemSolution')
```
