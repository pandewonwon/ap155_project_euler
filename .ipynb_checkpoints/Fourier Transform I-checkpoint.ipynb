```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N=2048
x=np.linspace(0,18*2*np.pi,N) #x-axis in normal space
deltaT=x[1]-x[0]

#frequencies
f1=1

#signal
y1=np.cos(2*np.pi*f1*x)

phi= 0.876 #phase shift
y2=np.cos((2*np.pi*f1*x)+phi) #phase shifted

#fourier transforms
freqs=np.fft.fftfreq(N,deltaT) #x-axis in frequency space

Yk1=np.fft.fft(y1) #FT
Yk1_mag=np.abs(Yk1) #magnitude
Yk1_phase=np.angle(Yk1)
Yk1_phase=np.unwrap(Yk1_phase) #phase

Yk2=np.fft.fft(y2)
Yk2_mag=np.abs(Yk2)
Yk2_phase=np.angle(Yk2)
Yk2_phase=np.unwrap(Yk2_phase)

#graphing
fig,axs=plt.subplots(1,3,figsize=(10,4))

plt.suptitle("Sinusoid signal and its FT")

axs[0].plot(x,y1,color="blue",label="cos(2*pi*f*x)")
axs[0].plot(x,y2,color="red", label="phase shifted +0.876")

axs[0].set_xlim(0,6)
axs[0].legend()

axs[0].set(
    xlabel="x",
    ylabel="y"
)


axs[1].plot(freqs,Yk1_mag,"b.", label="magnitude")
axs[1].plot(freqs,Yk2_mag,"r.", label="phase-shifted magnitude")

axs[1].set_xlim(0,2)
axs[1].legend()

axs[1].set(
    xlabel="f"
)
axs[1].grid()

axs[2].plot(freqs,Yk1_phase,"b.", label="phase")
axs[2].plot(freqs,Yk2_phase,"r.", label="phase-shifted")

axs[2].legend()

axs[2].set(
    xlabel="f"
)

plt.savefig("[FT] Sinusoid")
```

```python
N=2048
x=np.linspace(0,18*2*np.pi,N) #x-axis in normal space
deltaT=x[1]-x[0]

#frequencies
f1=1
f2=2

#signal
y1=np.cos(2*np.pi*f1*x)
y2=np.cos(2*np.pi*f2*x)

y=y1+y2 #product of signals

#fourier transforms
freqs=np.fft.fftfreq(N,deltaT) #x-axis in frequency space

Yk1=np.fft.fft(y1) #FT
Yk1_mag=np.abs(Yk1) #magnitude
Yk1_phase=np.angle(Yk1)
Yk1_phase=np.unwrap(Yk1_phase) #phase

Yk2=np.fft.fft(y2) #FT
Yk2_mag=np.abs(Yk2) #magnitude
Yk2_phase=np.angle(Yk2)
Yk2_phase=np.unwrap(Yk2_phase) #phase

Yk=np.fft.fft(y) #FT
Yk_mag=np.abs(Yk) #magnitude
Yk_phase=np.angle(Yk)
Yk_phase=np.unwrap(Yk_phase) #phase

#graphing
fig,axs=plt.subplots(1,3,figsize=(12,4))

plt.suptitle("Sum of sinusoid signals with different frequencies")

axs[0].plot(x,y1,color="blue",label="y1 with f=1")
axs[0].plot(x,y2,color="red", label="y2 with f=2")
axs[0].plot(x,y,color="purple", label="y1+y2")

axs[0].set_xlim(0,4)
axs[0].legend()

axs[0].set(
    xlabel="x",
    ylabel="y"
)

axs[1].plot(freqs,Yk1_mag,"b.", label="Y1 magnitude")
axs[1].plot(freqs,Yk2_mag,"r.", label="Y2 magnitude")
axs[1].plot(freqs,Yk_mag,".", color="violet", label="Y1+Y2 magnitude")

axs[1].set_xlim(0,2.5)
axs[1].legend()

axs[1].set(
    xlabel="f"
)
axs[1].grid()

axs[2].plot(freqs,Yk1_phase,"b.", label="Y1 phase")
axs[2].plot(freqs,Yk2_phase,"r.", label="Y2 phase")
axs[2].plot(freqs,Yk_phase,".", color="violet", label="Y1+Y2 phase")

axs[2].legend()

axs[2].set(
    xlabel="f"
)

plt.savefig("[FT] Sum of sinusoids")
```

```python
N=2048
x=np.linspace(0,20,N) #x-axis in normal space
deltaT=x[1]-x[0]

#signal
y1=(1*(x>3)+1*(x<7))-1 #large width
y2=(1*(x>3)+1*(x<4))-1 #small width

#fourier transforms
freqs=np.fft.fftfreq(N,deltaT) #x-axis in frequency space

Yk1=np.fft.fft(y1)
Yk1_mag=np.abs(Yk1)
Yk1_phase=np.angle(Yk1)
Yk1_phase=np.unwrap(Yk1_phase)

Yk2=np.fft.fft(y2)
Yk2_mag=np.abs(Yk2)
Yk2_phase=np.angle(Yk2)
Yk2_phase=np.unwrap(Yk2_phase)

#graphing
fig,axs=plt.subplots(1,3,figsize=(12,4))

#plt.suptitle("Square wave (Single)")

axs[0].plot(x,y1,color="blue", label="large width")
axs[0].plot(x,y2,color="red", label="small width")
axs[0].set_title("Single square wave")
axs[0].legend()

axs[0].set(
    xlabel="x",
    ylabel="y"
)

axs[1].plot(freqs,Yk1_mag, "b.", label="large width")
axs[1].plot(freqs, Yk2_mag, "r.", label="small width")
axs[1].set_title("FT magnitude")
axs[1].legend()
axs[1].set_xlim(0.1,10)
axs[1].set_ylim(0,30)

axs[1].set(
    xlabel="f"
)
axs[2].plot(freqs,Yk1_phase,"b.", label="large width")
axs[2].plot(freqs,Yk2_phase,"r.", label="small width")
axs[2].legend()
axs[2].set_title("FT phase")

axs[2].set(
    xlabel="f"
)

plt.savefig("[FT] Single square wave")
```

```python
N=2048
x=np.linspace(0,18*2*np.pi,N) #x-axis in normal space
deltaT=x[1]-x[0]

#frequencies
f1=1
f2=2

#signal
y1=np.sign(np.sin(2*np.pi*f1*x))
y2=np.sign(np.sin(2*np.pi*f2*x))

#fourier transforms
freqs=np.fft.fftfreq(N,deltaT) #x-axis in frequency space

Yk1=np.fft.fft(y1)
Yk1_mag=np.abs(Yk1)
Yk1_phase=np.angle(Yk1)
Yk1_phase=np.unwrap(Yk1_phase)

#graphing
fig,axs=plt.subplots(1,4,figsize=(16,4))

plt.suptitle("Square wave (Periodic)")

axs[0].plot(x,y1,color="blue", label="y1 with f=1")
axs[0].plot(x,y2,color="red", label="y2 with f=2")
axs[0].legend()
axs[0].set_xlim(0,4)
axs[0].set(
    xlabel="x",
    ylabel="y"
)


axs[1].plot(freqs,Yk1_mag, "b.", label="Y1 magnitude")
axs[1].plot(freqs,Yk2_mag, "r.", label="Y2 magnitude")
axs[1].legend()
axs[1].set_xlim(0,10)
#axs[1].set_ylim(0,1000)
axs[1].set(
    xlabel="f"
)

axs[2].plot(freqs,Yk1_phase, "b.", label="Y1 Phase")
axs[2].legend()
#axs[2].plot(freqs,Yk2_phase, "r.")
#axs[2].set_ylim(-2000,500)
axs[2].set(
    xlabel="f"
)
axs[1].grid()

axs[3].plot(freqs, Yk2_phase,"r.", label="Y2 Phase")
axs[3].legend()
axs[3].set(
    xlabel="f"
)

plt.savefig("[FT] Periodic square wave")
```

```python
N=2048
t=np.linspace(0,20,N) #x-axis in normal space
deltaT=t[1]-t[0]

#frequency
f=1
T=1/f

#signal
y=(t%T)/T

#fourier transforms
freqs=np.fft.fftfreq(N,deltaT) #x-axis in frequency space

Yk=np.fft.fft(y)
Yk_mag=np.abs(Yk)
Yk_phase=np.angle(Yk)
Yk_phase=np.unwrap(Yk_phase)

#graphing
fig,axs=plt.subplots(1,3,figsize=(12,4))

#plt.suptitle("Square wave (Single)")

axs[0].plot(x,y,color="blue")
axs[0].set_xlim(0,40)
axs[0].set_title("Sawtooth")

axs[0].set(
    xlabel="x",
    ylabel="y"
)

axs[1].plot(freqs,Yk_mag, "b.")
axs[1].set(
    xlabel="f"
)
axs[1].set_title("FT Magnitude")
axs[1].set_xlim(0,40)
axs[1].grid()

axs[2].plot(freqs, Yk_phase, "b.")
axs[2].set(
    xlabel="f"
)
axs[2].set_title("FT Phase")

plt.savefig("[FT] Sawtooth")
```

```python
#ramp
N=2048
t=np.linspace(0,20,N) #x-axis in normal space
deltaT=t[1]-t[0]

#frequency
f=1
T=1/f

#signal
y= (2 * np.abs(2 * (t / T - np.floor(t / T + 0.5))))-1

#fourier transforms
freqs=np.fft.fftfreq(N,deltaT) #x-axis in frequency space

Yk=np.fft.fft(y)
Yk_mag=np.abs(Yk)
Yk_phase=np.angle(Yk)
Yk_phase=np.unwrap(Yk_phase)

#graphing
fig,axs=plt.subplots(1,3,figsize=(12,4))

#plt.suptitle("Square wave (Single)")

axs[0].plot(x,y,color="blue")
axs[0].set_xlim(0,20)
axs[0].set_title("Ramp (Triangle Wave)")
axs[0].set(
    xlabel="x",
    ylabel="y"
)

axs[1].plot(freqs,Yk_mag, "b.")
axs[1].set(
    xlabel="f"
)
axs[1].set_title("FT Magnitude")
axs[1].grid()
axs[1].set_xlim(0,40)

axs[2].plot(freqs, Yk_phase, "b.")
axs[2].set(
    xlabel="f"
)
axs[2].set_title("FT Phase")

plt.savefig("[FT] Ramp Triangle wave")
```

```python
N=2048
x = np.linspace(0, 10, N)
deltaT=x[1]-x[0]

amplitude = 1.0
mean = 5

#sigma
std_dev1 = 0.5
std_dev2=2

#signal
y1 = amplitude * np.exp(-((x - mean)**2 / (2 * std_dev1**2)))
y2 = amplitude * np.exp(-((x - mean)**2 / (2 * std_dev2**2)))

#Fourier transform
freqs=np.fft.fftfreq(N,deltaT)

Yk1=np.fft.fft(y1)
Yk1_mag=np.abs(Yk1)
Yk1_phase=np.angle(Yk1)
Yk1_phase=np.unwrap(Yk1_phase)

Yk2=np.fft.fft(y2)
Yk2_mag=np.abs(Yk2)
Yk2_phase=np.angle(Yk2)
Yk2_phase=np.unwrap(Yk2_phase)

#graphing
fig,axs=plt.subplots(1,4,figsize=(16,4))

plt.suptitle("Gaussian signal and its FT")

axs[0].plot(x,y1,color="blue", label="y1 with sigma=0.5")
axs[0].plot(x,y2, color="red", label="y2 with sigma=2")
axs[0].legend()
axs[0].set(
    xlabel="x",
    ylabel="y"
)

axs[1].plot(freqs,Yk1_mag, "b.", label="Y1 magnitude")
axs[1].plot(freqs,Yk2_mag, "r.", label="Y2 magnitude")
axs[1].legend()
axs[1].set(
    xlabel="f"
)
axs[1].grid()
axs[1].set_xlim(-1,1)

axs[2].plot(freqs, Yk1_phase, "b.", label="Y1 phase")
axs[2].legend()
axs[2].set(
    xlabel="f"
)

axs[3].plot(freqs, Yk2_phase, "r.", label="Y2 phase")
axs[3].legend()
axs[3].set(
    xlabel="f"
)

plt.savefig("[FT] Gaussian")
```

```python
#FINITE TIME SINUSOID
N=2048
x=np.linspace(0,20,N) #x-axis in normal space
deltaT=x[1]-x[0]

#frequency
f=1

#signal
y1=(1*(x>3)+1*(x<7))-1 #square wave
y2=np.cos(2*np.pi*f*x) #sinusoid

y=y1*y2 #finite-time sinusoid

#fourier transforms
freqs=np.fft.fftfreq(N,deltaT) #x-axis in frequency space

Yk=np.fft.fft(y)
Yk_mag=np.abs(Yk)
Yk_phase=np.angle(Yk)
Yk_phase=np.unwrap(Yk_phase)

#graphing
fig,axs=plt.subplots(1,3,figsize=(12,4))

plt.suptitle("Finite-time sinusoid")

axs[0].plot(x,y1, color="blue", linestyle="--", label="square wave")
axs[0].plot(x,y2, color="red", linestyle="--", label="sinusoid")
axs[0].plot(x,y, color="violet", label="finite-time sinusoid")
axs[0].set_xlim(0,10)
axs[0].legend()
#axs[0].set_title("Finite-time sinusoid")
axs[0].set(
    xlabel="x",
    ylabel="y"
)

axs[1].plot(freqs,Yk_mag,".", color="violet", label="FT magnitude")
axs[1].set_xlim(0,5)
axs[1].legend()
axs[1].grid()
#axs[1].set_title("FT magnitude")
axs[1].set(
    xlabel="f"
)

axs[2].plot(freqs,Yk_phase, ".", color="violet", label="FT Phase")
axs[2].legend()
#axs[2].set_title("FT phase")
axs[2].set(
    xlabel="f"
)

plt.savefig("[FT] Finite-time sinusoid")
```

```python
#Gaussian wave packet
N=2048
x = np.linspace(0, 10, N)
deltaT=x[1]-x[0]

amplitude = 1.0
mean = 5

#sigma
std_dev=1

#frequency
f=1

#signal
y1 = amplitude * np.exp(-((x - mean)**2 / (2 * std_dev**2))) #Gaussian wave
y2 =np.cos(2*np.pi*f*x) #sinusoid

y=y1*y2

#Fourier transform
freqs=np.fft.fftfreq(N,deltaT)

Yk=np.fft.fft(y)
Yk_mag=np.abs(Yk)
Yk_phase=np.angle(Yk)
Yk_phase=np.unwrap(Yk_phase)

#graphing
fig,axs=plt.subplots(1,3,figsize=(12,4))

plt.suptitle("Gaussian wave packet and its FT")

axs[0].plot(x,y1,color="blue", linestyle="--", label="gaussian wave")
axs[0].plot(x,y2, color="red", linestyle="--", label="sinusoid")
axs[0].plot(x,y, color="violet", label="gaussian wave packet")
axs[0].legend()
axs[0].set(
    xlabel="x",
    ylabel="y"
)

axs[1].plot(freqs,Yk_mag, ".", color="violet", label="FT magnitude")
axs[1].set_xlim(0,4)
axs[1].grid()
axs[1].legend()
axs[1].set(
    xlabel="f"
)

axs[2].plot(freqs,Yk_phase, ".", color="violet", label="FT Phase")
axs[2].legend()
axs[2].set(
    xlabel="f"
)

plt.savefig("[FT] Gaussian wave packet")
```
