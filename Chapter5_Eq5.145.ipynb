```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def phi(x):
    return (-x**5)+(4*x**4)-(4*x**3)+(x**2*np.exp(x))-(2*x**2)-(4*x*np.exp(x))+(8*x)+(4*np.exp(x))-8

def phi_prime(x):
    return (-5*x**4)+(16*x**3)-(12*x**2)+(x**2*np.exp(x))+(2*x*np.exp(x))-(4*x)-(4*x*np.exp(x))+8

def phi_dprime(x):
    return (-20*x**3)+(48*x**2)-(24*x)+(x**2*np.exp(x))+(2*x*np.exp(x))+(2*x*np.exp(x))+(2*np.exp(x))-4-(4*x*np.exp(x))-(4*np.exp(x))
```

```python
#Graphing phi and its derivatives
x=np.linspace(-3,5,100)

fig,axs=plt.subplots(1,3,figsize=(10,4))
ax1,ax2,ax3=axs

ax1.plot(x,phi(x))
ax1.set_title("phi(x)")
ax1.axhline(0,linestyle="--",color="r")
ax1.grid()

ax2.plot(x,phi_prime(x))
ax2.set_title("First derivative of phi(x)")
ax2.axhline(0,linestyle="--",color="r")
ax2.grid()

ax3.plot(x,phi_dprime(x))
ax3.set_title("Second derivative of phi(x)")
ax3.axhline(0,linestyle="--",color="r")
ax3.grid()
```

```python
def golden(phi,x0,x1,kmax=200,tol=1.e-8):
    varphi=0.5*(1+np.sqrt(5)) #Golden Ratio
    for k in range(1,kmax):
        x2=x0+(x1-x0)/(varphi+1)
        x3=x0+(x1-x0)*varphi/(varphi+1)

        if phi(x3)<phi(x2):
            x0=x2
        else:
            x1=x3

        xnew=(x0+x1)/2
        xdiff=abs(x1-x0)
        #rowf="{0:2d} {1:1.16f} {2:1.16f} {3:1.16f}"
        #print(rowf.format(k,xnew,xdiff,phi(xnew)))

        if abs(xdiff)<tol:
            break
    else:
        xnew=None
    return xnew
```

```python
Roots=golden(phi,-1.5,0) #local minima
print("local minima at", Roots)
```

```python
Roots=golden(phi,3.5,4.5) #local minima
print("local minima at", Roots)
```
